```python
import os, re, glob, zipfile, gc
import numpy as np
import pandas as pd
import torch
from IPython.display import display

# ---------- Settings you can change ----------
CHUNK_SIZE = 5          # how many sentences to join before translating (task says 5)
W_NSP      = 1.0        # importance of the BERT "next sentence" signal
W_EMB      = 1.0        # importance of the sentence-meaning similarity signal
W_TFIDF    = 1.0        # importance of the shared-words signal
LIMIT_GB   = 1.5        # GPU memory limit from the project rules
# ---------------------------------------------

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)
if device == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("No GPU found - everything will still work, just slower. (Colab: Runtime > Change runtime type > GPU)")

memory_log = {}   # we store the peak GPU memory of every step here

def free_gpu():
    """Remove unused things from (GPU) memory."""
    gc.collect()
    if device == "cuda":
        torch.cuda.empty_cache()

def report_gpu(step_name):
    """Print the peak GPU memory used since the last report, then reset the counter."""
    if device == "cuda":
        peak = torch.cuda.max_memory_allocated() / 1024**3
        memory_log[step_name] = round(peak, 3)
        status = "OK (under limit)" if peak < LIMIT_GB else "TOO HIGH!"
        print(f"[GPU memory] {step_name}: peak {peak:.2f} GB / limit {LIMIT_GB} GB -> {status}")
        torch.cuda.reset_peak_memory_stats()
    else:
        memory_log[step_name] = 0.0
        print(f"[GPU memory] {step_name}: running on CPU, GPU memory used = 0 GB")
```

```python
ZIP_NAME = "Sample-Raw-No-Punctuations.zip"

# If the zip is not in the notebook folder, ask for an upload (works on Google Colab)
if not os.path.exists(ZIP_NAME):
    try:
        from google.colab import files
        print("Please choose the file 'Sample-Raw-No-Punctuations.zip' ...")
        uploaded = files.upload()
        ZIP_NAME = list(uploaded.keys())[0]
    except ImportError:
        raise FileNotFoundError(f"Put '{ZIP_NAME}' in the same folder as this notebook and run this cell again.")

with zipfile.ZipFile(ZIP_NAME) as z:
    z.extractall("data")

def file_number(path):
    # sorts sample-1, sample-2, ... sample-10 in the right order
    m = re.search(r"(\d+)", os.path.basename(path))
    return int(m.group(1)) if m else 0

txt_files = [p for p in glob.glob("data/**/*.txt", recursive=True)
             if "__MACOSX" not in p and not os.path.basename(p).startswith("._")]
txt_files = sorted(txt_files, key=file_number)
assert len(txt_files) > 0, "No .txt files found inside the zip!"

raw_texts = {}
for path in txt_files:
    name = os.path.splitext(os.path.basename(path))[0]      # e.g. 'sample-1'
    with open(path, encoding="utf-8-sig") as f:
        raw_texts[name] = f.read()

print("Files found:", list(raw_texts.keys()))
```

```python
def clean_text(text):
    # 1) remove invisible characters
    for ch in ["\u200d", "\u200c", "\ufeff"]:
        text = text.replace(ch, "")
    # 2) ignore the NULL placeholder
    text = text.replace("NULL", " ")
    # 3) make spaces tidy
    return " ".join(text.split())

clean_texts = {name: clean_text(t) for name, t in raw_texts.items()}

# ---------- small analysis table ----------
rows = []
for name in raw_texts:
    raw, cln = raw_texts[name], clean_texts[name]
    words = cln.split()
    rows.append({
        "file": name,
        "characters (raw)": len(raw),
        "NULL tokens": raw.count("NULL"),
        "words (clean)": len(words),
        "unique words": len(set(words)),
        "English words": len(re.findall(r"[A-Za-z]{2,}", cln)),
        "numbers": len(re.findall(r"\b\d+\b", cln)),
    })
display(pd.DataFrame(rows))

# ---------- what is around NULL? ----------
first = list(raw_texts)[0]
tokens = raw_texts[first].split()
print("\nExamples of NULL in", first, "(words before [NULL] words after):")
shown = 0
for i, w in enumerate(tokens):
    if w == "NULL" and shown < 4:
        print("  ", " ".join(tokens[max(0, i-3):i]), "[NULL]", " ".join(tokens[i+1:i+4]))
        shown += 1

print("\nFirst 300 characters of cleaned", first, ":\n", clean_texts[first][:300])
```

```python
# ---------- helper functions ----------
PUNCT_CHARS = ",.।?!;:\"'()-"
QUESTION_WORDS = {"क्या","कौन","कब","क्यों","कैसे","कैसा","कैसी","कहाँ","कहां","किसने","किसे","किसको",
                  "किसका","किसकी","किधर","कितना","कितनी","कितने"}
EXCLAIM_START = {"अरे","ओह","वाह","हाय","शाबाश","काश","ओए","अबे","उफ़","उफ","अहा","हे","छि","अरी","ऐ"}
COMMAND_WORDS = {"जाओ","भागो","रुको","चुप","सुनो","देखो","आओ","चलो","बचाओ","निकल","ठहरो","बैठो",
                 "उठो","छोड़ो","बस","खबरदार"}

def fix_ending(sentence):
    """Choose the ending mark: ? or ! or । (Hindi full stop)."""
    s = sentence.strip().rstrip(".।").rstrip()
    if not s:
        return s
    if s[-1] in "?!":                       # the model already chose ? or !, keep it
        return s
    words = [w.strip(PUNCT_CHARS) for w in s.split()]
    words = [w for w in words if w]
    if not words:
        return s + "।"
    # Question: has a question word (but not inside an indirect "... कि ..." part)
    for k, w in enumerate(words):
        if w in QUESTION_WORDS and "कि" not in words[:k]:
            return s + "?"
    # Question tag at the end: "... है ना", "... हो न"
    if len(words) >= 2 and words[-1] in {"ना", "न"} and words[-2] in {"है","हैं","हो","था","थी","होगा"}:
        return s + "?"
    # Exclamation: starts with an interjection, or is a short command
    if words[0] in EXCLAIM_START:
        return s + "!"
    if len(words) <= 4 and any(w in COMMAND_WORDS for w in words):
        return s + "!"
    return s + "।"

# words that usually END a Hindi sentence (only used by the backup splitter)
END_WORDS = {"है","हैं","था","थे","थी","थीं","गया","गए","गई","गयी","होगा","होगी","होंगे","रहा","रहे","रही",
             "सकता","सकती","सकते","चाहिए","जाता","जाती","जाते","हुआ","हुई","हुए","दिया","किया","लिया"}
# if the NEXT word is one of these, the sentence has not ended yet
NOT_START_WORDS = {"कि","तो","और","तथा","या","भी","ही","है","हैं","था","थे","थी","थीं","गया","गए","गई","गयी",
                   "होगा","होगी","होंगे","रहा","रहे","रही","हो","होता","होती","होते","सकता","सकती","सकते",
                   "जाता","जाती","जाते","चुका","चुके","चुकी","करता","करती","करते","गये","जा","दिया","लिया",
                   "किया","पड़ा","पड़ी","पड़े","वाला","वाली","वाले",""}

def rule_based_sentences(text, min_words=5):
    """Simple BACKUP splitter (used only if the pretrained model is unavailable)."""
    words, sentences, current = text.split(), [], []
    for k, w in enumerate(words):
        current.append(w)
        nxt = words[k + 1] if k + 1 < len(words) else ""
        if w in END_WORDS and nxt not in NOT_START_WORDS and len(current) >= min_words:
            sentences.append(" ".join(current))
            current = []
    if current:
        sentences.append(" ".join(current))
    return sentences

# ---------- load the pretrained punctuation model ----------
try:
    from punctuators.models import PunctCapSegModelONNX
    punct_model = PunctCapSegModelONNX.from_pretrained("pcs_47lang")
    USE_PRETRAINED_PUNCT = True
    print("Pretrained punctuation model loaded.")
except Exception as e:
    USE_PRETRAINED_PUNCT = False
    print("WARNING: could not load the pretrained punctuation model ->", e)
    print("Using the simple backup splitter instead (results will be less accurate).")

def add_punctuation(text):
    """Raw Hindi text (no punctuation) -> list of punctuated sentences."""
    sentences = None
    if USE_PRETRAINED_PUNCT:
        try:
            sentences = punct_model.infer([text], apply_sbd=True)[0]
        except Exception as e:
            print("WARNING: punctuation model failed on one file ->", e, "| using backup splitter for it.")
    if sentences is None:
        sentences = rule_based_sentences(text)
    return [fix_ending(s) for s in sentences if s.strip()]

# ---------- run on all files ----------
results = {name: {} for name in clean_texts}
for name, text in clean_texts.items():
    results[name]["sentences_jumbled"] = add_punctuation(text)
    print(f"{name}: {len(results[name]['sentences_jumbled'])} sentences")

print("\nFirst 6 punctuated sentences of", first, "(still in jumbled order):")
for s in results[first]["sentences_jumbled"][:6]:
    print(" -", s)
report_gpu("Step 4 punctuation")
```

```python
from transformers import AutoTokenizer, BertForNextSentencePrediction
from sentence_transformers import SentenceTransformer
from sklearn.feature_extraction.text import TfidfVectorizer

sentence_lists = {name: results[name]["sentences_jumbled"] for name in results}
nsp_scores, emb_scores, tfidf_scores = {}, {}, {}

# ======================= Signal 1: BERT next-sentence prediction =======================
NSP_NAME = "bert-base-multilingual-cased"
nsp_tok = AutoTokenizer.from_pretrained(NSP_NAME)
nsp_model, load_info = BertForNextSentencePrediction.from_pretrained(NSP_NAME, output_loading_info=True)
nsp_model = nsp_model.to(device).eval()
# safety check: make sure the "next sentence" part of the model was really loaded
NSP_READY = not any("seq_relationship" in k for k in load_info["missing_keys"])
print("NSP head loaded correctly:", NSP_READY)

# Hindi name -> English spelling. Add more names here if you spot any being changed.
GLOSSARY = {
    # places
    "गढ़ मुक्तेश्वर": "Garhmukteshwar", "मुक्तेश्वर": "Mukteshwar", "बल्लभपुर": "Ballabhpur",
    "बनारस छावनी": "Banaras Cantonment", "बनारस": "Banaras", "वाराणसी": "Varanasi", "काशी": "Kashi",
    "इलाहाबाद": "Allahabad", "प्रयाग": "Prayag", "अयोध्या": "Ayodhya", "मथुरा": "Mathura",
    "उज्जैन": "Ujjain", "द्वारका": "Dwarka", "हरिद्वार": "Haridwar", "कांचीपुरम": "Kanchipuram",
    "कांजीवरम": "Kanjeevaram", "चेंगलपट": "Chengalpattu", "अरकोनम": "Arakkonam", "चेन्नै": "Chennai",
    "भटनी": "Bhatni", "छपरा": "Chhapra", "मुगलसराय": "Mughalsarai", "लखनऊ": "Lucknow",
    "गाजियाबाद": "Ghaziabad", "मेरठ": "Meerut", "दिल्ली": "Delhi", "हस्तिनापुर": "Hastinapur",
    "तिजारा": "Tijara", "शिमला": "Shimla", "भोपाल": "Bhopal", "मुंबई": "Mumbai", "कोटा": "Kota",
    "आगरा": "Agra", "गोरखपुर": "Gorakhpur", "अमृतसर": "Amritsar", "देहरादून": "Dehradun",
    "लक्सर": "Laksar", "महू": "Mhow", "धनुष्कोटि": "Dhanushkodi", "पुष्कर": "Pushkar",
    "बद्रीनाथ": "Badrinath", "अलवर": "Alwar", "कश्मीरी गेट": "Kashmiri Gate", "खांडवी": "Khandavi",
    "नागपुर": "Nagpur", "मणिपुर": "Manipur", "गजपुर": "Gajapur", "पोदनपुर": "Podanpur",
    "असम": "Assam", "चीन": "China", "जापान": "Japan", "कर्नाटक": "Karnataka", "तमिलनाडु": "Tamil Nadu",
    "राजस्थान": "Rajasthan", "पंजाब": "Punjab", "हिमाचल": "Himachal", "बिहार": "Bihar",
    # rivers
    "गंगा": "Ganga", "यमुना": "Yamuna", "सरयू": "Sarayu",
    # people / gods
    "परशुराम": "Parashurama", "श्रीकृष्ण": "Shri Krishna", "कृष्ण": "Krishna", "राम": "Ram",
    "शिवि": "Shivi", "युधिष्ठिर": "Yudhishthira", "अर्जुन": "Arjuna", "महाकाल": "Mahakal",
    "महाभारत": "Mahabharata", "महावीर": "Mahavir", "आदिनाथ": "Adinath", "शांतिनाथ": "Shantinath",
    "कुंथनाथ": "Kunthunath", "अरहनाथ": "Arahnath", "मल्लिनाथ": "Mallinath", "पार्श्वनाथ": "Parshvanath",
    "श्रेयंषकुमार": "Shreyanshkumar", "सरस्वती": "Saraswati", "बिहारी लाल": "Bihari Lal",
    "कैलाशनाथ": "Kailashnath", "चौधरी": "Chaudhary", "राहुल": "Rahul", "अंजलि": "Anjali",
    "अमन": "Aman", "टीना": "Tina", "रिफ़त": "Rifat", "राजेश": "Rajesh",
}

def protect_names(sentence):
    """Replace Hindi names by their English spelling (longest names first)."""
    for hindi in sorted(GLOSSARY, key=len, reverse=True):
        sentence = re.sub(r"(?<!\S)" + re.escape(hindi) + r"(?=[\s।,?!;:]|$)", GLOSSARY[hindi], sentence)
    return sentence

@torch.no_grad()
def nsp_matrix(sentences):
    """M[i][j] = log-probability that sentence j comes right after sentence i."""
    n = len(sentences)
    M = np.zeros((n, n), dtype=np.float32)
    for i in range(n):
        others = [j for j in range(n) if j != i]
        enc = nsp_tok([sentences[i]] * len(others), [sentences[j] for j in others],
                      return_tensors="pt", padding=True, truncation=True, max_length=128).to(device)
        logits = nsp_model(**enc).logits                       # shape: (len(others), 2)
        log_prob_next = torch.log_softmax(logits, dim=-1)[:, 0]  # index 0 = "B is the next sentence"
        M[i, others] = log_prob_next.float().cpu().numpy()
    return M

if NSP_READY:
    for name, sents in sentence_lists.items():
        nsp_scores[name] = nsp_matrix(sents)
        print("NSP done:", name)
else:
    print("WARNING: NSP part missing in this checkpoint -> only signals 2 and 3 will be used.")

del nsp_model, nsp_tok
free_gpu()
report_gpu("Step 5a BERT next-sentence model")

# ======================= Signal 2: sentence-meaning similarity =======================
EMB_NAME = "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
emb_model = SentenceTransformer(EMB_NAME, device=device)

def embedding_matrix(sentences):
    emb = emb_model.encode(sentences, convert_to_numpy=True, normalize_embeddings=True,
                           batch_size=32, show_progress_bar=False)
    emb = emb.astype(np.float32)
    return emb @ emb.T            # cosine similarity (vectors are already normalised)

for name, sents in sentence_lists.items():
    emb_scores[name] = embedding_matrix(sents)
    print("Embeddings done:", name)

del emb_model
free_gpu()
report_gpu("Step 5b sentence-embedding model")

# ======================= Signal 3: shared words (TF-IDF) =======================
def tfidf_matrix(sentences):
    plain = [re.sub(r"[।,?!.;:\"'()\-]", " ", s) for s in sentences]
    vec = TfidfVectorizer(analyzer="char_wb", ngram_range=(2, 4), sublinear_tf=True)
    X = vec.fit_transform(plain)
    return (X @ X.T).toarray().astype(np.float32)

for name, sents in sentence_lists.items():
    tfidf_scores[name] = tfidf_matrix(sents)
print("TF-IDF done for all files.")
```

```python
def zscore(M):
    """Standardise a score matrix (ignoring the diagonal) so different signals can be added fairly."""
    n = M.shape[0]
    off_diag = ~np.eye(n, dtype=bool)
    mean, std = M[off_diag].mean(), M[off_diag].std()
    if std < 1e-9:
        std = 1.0
    Z = (M - mean) / std
    np.fill_diagonal(Z, 0.0)
    return Z

def path_cost(path, cost):
    return sum(cost[path[k]][path[k + 1]] for k in range(len(path) - 1))

def greedy_path(start, cost):
    """Start at one sentence, always jump to the best unused next sentence."""
    n = len(cost)
    path, left = [start], set(range(n))
    left.remove(start)
    while left:
        cur = path[-1]
        nxt = min(left, key=lambda j: cost[cur][j])
        path.append(nxt)
        left.remove(nxt)
    return path

def improve_path(path, cost):
    """Try moving small blocks (1-4 sentences) to other places while it makes the total better."""
    best, best_c, n = path[:], path_cost(path, cost), len(path)
    while True:
        found = False
        for seg_len in (1, 2, 3, 4):
            for i in range(n - seg_len + 1):
                block = best[i:i + seg_len]
                rest = best[:i] + best[i + seg_len:]
                for j in range(len(rest) + 1):
                    if j == i:
                        continue
                    candidate = rest[:j] + block + rest[j:]
                    c = path_cost(candidate, cost)
                    if c < best_c - 1e-9:
                        best, best_c, found = candidate, c, True
                        break
                if found: break
            if found: break
        if not found:
            break
    return best

def find_best_order(score, n_starts=5):
    """score[i][j] = how good is 'j after i'. Returns list of sentence indexes in best order."""
    n = score.shape[0]
    if n <= 2:
        return list(range(n))
    cost = (-score).tolist()                      # lower cost = better
    starts = []
    for s in range(n):
        p = greedy_path(s, cost)
        starts.append((path_cost(p, cost), p))
    starts.sort(key=lambda x: x[0])
    best_path, best_c = None, float("inf")
    for _, p in starts[:n_starts]:                # polish the 5 best greedy paths
        p = improve_path(p, cost)
        c = path_cost(p, cost)
        if c < best_c:
            best_path, best_c = p, c
    return best_path

for name, sents in sentence_lists.items():
    total = W_EMB * zscore(emb_scores[name]) + W_TFIDF * zscore(tfidf_scores[name])
    if name in nsp_scores:
        total = total + W_NSP * zscore(nsp_scores[name])
    order = find_best_order(total)
    results[name]["order_indexes"] = order
    results[name]["sentences_ordered"] = [sents[i] for i in order]
    print(f"{name}: ordered {len(order)} sentences")

print("\nFirst 6 sentences of", first, "after ordering:")
for s in results[first]["sentences_ordered"][:6]:
    print(" -", s)
```

```python
from transformers import AutoModelForSeq2SeqLM

MT_NAME = "Helsinki-NLP/opus-mt-hi-en"
mt_tok = AutoTokenizer.from_pretrained(MT_NAME)
mt_model = AutoModelForSeq2SeqLM.from_pretrained(MT_NAME).to(device).eval()

@torch.no_grad()
def translate(sentences, batch_size=8):
    """List of Hindi sentences -> list of English sentences."""
    english = []
    for i in range(0, len(sentences), batch_size):
        batch = sentences[i:i + batch_size]
        enc = mt_tok(batch, return_tensors="pt", padding=True, truncation=True, max_length=256).to(device)
        out = mt_model.generate(**enc, num_beams=4, max_length=256)
        english.extend(mt_tok.batch_decode(out, skip_special_tokens=True))
    return english

def make_chunks(sentences, size):
    return [sentences[i:i + size] for i in range(0, len(sentences), size)]

for name in results:
    hindi_chunks = make_chunks(results[name]["sentences_ordered"], CHUNK_SIZE)
    english_chunks = []
    for chunk in hindi_chunks:
        english_sentences = translate([protect_names(s) for s in chunk])
        english_chunks.append(" ".join(e.strip() for e in english_sentences))
    results[name]["chunks_hi"] = [" ".join(c) for c in hindi_chunks]
    results[name]["chunks_en"] = english_chunks
    print(f"{name}: {len(hindi_chunks)} chunks translated")

del mt_model, mt_tok
free_gpu()
report_gpu("Step 6 translation model")

print("\n=== Example: first chunk of", first, "===")
print("HINDI  :", results[first]["chunks_hi"][0])
print("ENGLISH:", results[first]["chunks_en"][0])
```
